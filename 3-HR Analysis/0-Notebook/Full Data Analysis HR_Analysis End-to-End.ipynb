# Phase 3 — Data Quality Assessment & Cleaning

This phase converts the raw, 1,480-row extract into a clean, analysis-ready dataset at the verified grain of **1,470 unique employees**.

Each step below documents the specific data quality issue found, the evidence for it, and the exact rule used to resolve it — so the cleaning logic is auditable rather than a black box.

The steps, in order:
1. Duplicate resolution (including the 3 conflicting pairs identified in Phase 2)
2. Missing value treatment (`YearsWithCurrManager`)
3. Categorical value consistency (`BusinessTravel`)
4. Logical consistency re-validation
5. Zero-variance column removal
6. Outlier assessment (flagged, not removed)


## 1. Duplicate Resolution

Phase 2 found 10 duplicate `EmployeeNumber` pairs. Re-verifying against the full row (all columns except `EmpID`, since `EmpID` itself repeats identically within each pair) confirms:

- **7 pairs are true full-record duplicates** — safe to drop the second occurrence.
- **3 pairs (EmployeeNumber 2060, 2061, 2062) disagree on `YearsWithCurrManager`** — these are not full duplicates and need an explicit resolution rule, not a blind `keep='first'`.

**Resolution rule:** for each conflicting pair, keep whichever value satisfies `YearsWithCurrManager <= YearsAtCompany` (a value with current manager can never exceed total tenure at the company). If both values are logically valid, keep the higher one as the more conservative assumption. This rule matters in practice — for EmployeeNumber 2060 specifically, the *higher* of the two values (5.0) actually **violates** the constraint (`YearsAtCompany` = 4), so a naive "always keep the max" rule would have kept the wrong value for this employee.

```python
# Identify duplicate EmployeeNumbers
dup_mask = df.duplicated(subset=['EmployeeNumber'], keep=False)
dup_ids = df.loc[dup_mask, 'EmployeeNumber'].unique()
print(f"Duplicate EmployeeNumbers found: {len(dup_ids)}")

# Resolve each pair: drop true full-duplicates' second row;
# for conflicting YearsWithCurrManager pairs, keep the logically valid value
rows_to_drop = []
conflict_log = []

for empnum in dup_ids:
    idx = df[df['EmployeeNumber'] == empnum].index
    r0, r1 = df.loc[idx[0]], df.loc[idx[1]]

    if r0.drop('EmpID').equals(r1.drop('EmpID')):
        # True full-record duplicate -> drop the second occurrence
        rows_to_drop.append(idx[1])
    else:
        # Conflicting YearsWithCurrManager -> resolve using the logical constraint
        v0, v1 = r0['YearsWithCurrManager'], r1['YearsWithCurrManager']
        yac = r0['YearsAtCompany']
        v0_valid = pd.isna(v0) or v0 <= yac
        v1_valid = pd.isna(v1) or v1 <= yac

        if v0_valid and not v1_valid:
            keep_idx, drop_idx = idx[0], idx[1]
        elif v1_valid and not v0_valid:
            keep_idx, drop_idx = idx[1], idx[0]
        else:
            # both valid -> keep the higher value
            keep_idx = idx[0] if v0 >= v1 else idx[1]
            drop_idx = idx[1] if keep_idx == idx[0] else idx[0]

        rows_to_drop.append(drop_idx)
        conflict_log.append((empnum, v0, v1, yac, df.loc[keep_idx, 'YearsWithCurrManager']))

df_clean = df.drop(index=rows_to_drop).reset_index(drop=True)

print(f"\nRows dropped: {len(rows_to_drop)}")
print(f"Shape after deduplication: {df_clean.shape} (expected 1,470 rows)")
print("\nConflicting-pair resolutions (EmployeeNumber, value_A, value_B, YearsAtCompany, value_kept):")
for row in conflict_log:
    print(f"  {row}")
```

## 2. Missing Value Treatment — `YearsWithCurrManager`

57 rows (3.85%) have a missing `YearsWithCurrManager` value. Before choosing an imputation method, two checks:

- **Are the missing values concentrated among brand-new hires** (`YearsAtCompany` = 0, where 0 would be a legitimate default rather than "missing")? No — only 2 of the 57 have `YearsAtCompany` = 0; the missing values span the full tenure range (0 to 22 years), so this is genuine missing data, not a default-value pattern.
- **Is there a variable strongly correlated with `YearsWithCurrManager`** that can support a reasonable imputation? Yes — correlation with `YearsAtCompany` is **0.76**, and every `YearsAtCompany` value present among the missing rows has at least 10 non-null observations to compute a median from.

**Method used:** impute with the median `YearsWithCurrManager` within the same `YearsAtCompany` value (not a single global median, which would ignore tenure and bias toward the middle of the distribution for both new and long-tenured employees). A safety cap enforces `YearsWithCurrManager <= YearsAtCompany` after imputation, consistent with the constraint used above.

```python
null_before = df_clean['YearsWithCurrManager'].isnull().sum()
print(f"Nulls before imputation: {null_before}")

# Group-median imputation (grouped by YearsAtCompany), with a logical safety cap
group_median = df_clean.groupby('YearsAtCompany')['YearsWithCurrManager'].transform('median')
df_clean['YearsWithCurrManager'] = df_clean['YearsWithCurrManager'].fillna(group_median)
df_clean['YearsWithCurrManager'] = df_clean[['YearsWithCurrManager', 'YearsAtCompany']].min(axis=1)

print(f"Nulls after imputation: {df_clean['YearsWithCurrManager'].isnull().sum()}")
print(f"Total nulls remaining in dataset: {df_clean.isnull().sum().sum()}")
```

## 3. Categorical Value Consistency — `BusinessTravel`

Phase 2's `nunique()` summary reported 4 unique values for `BusinessTravel`, which was taken at face value as 4 legitimate categories. A closer look at the actual value counts shows this is wrong:

| Value | Count |
|---|---|
| Travel_Rarely | 1,036 |
| Travel_Frequently | 277 |
| Non-Travel | 150 |
| TravelRarely | 7 |

`TravelRarely` (no underscore) is not a fourth category — it's a formatting inconsistency for `Travel_Rarely`, affecting 7 records. This is exactly the kind of issue a raw `nunique()` count can hide, since it counts distinct strings, not distinct real-world categories.

```python
print("BusinessTravel before standardization:")
print(df_clean['BusinessTravel'].value_counts())

df_clean['BusinessTravel'] = df_clean['BusinessTravel'].replace({'TravelRarely': 'Travel_Rarely'})

print("\nBusinessTravel after standardization:")
print(df_clean['BusinessTravel'].value_counts())
```

## 4. Logical Consistency Re-validation

Re-running the tenure logical checks after deduplication and imputation confirms every fix actually resolved the underlying issue, rather than just suppressing the symptom:
- `YearsInCurrentRole` should never exceed `YearsAtCompany`
- `YearsSinceLastPromotion` should never exceed `YearsAtCompany`
- `YearsWithCurrManager` should never exceed `YearsAtCompany`
- `YearsAtCompany` should never exceed `TotalWorkingYears`

```python
logic_checks = {
    'YearsInCurrentRole > YearsAtCompany': (df_clean['YearsInCurrentRole'] > df_clean['YearsAtCompany']).sum(),
    'YearsSinceLastPromotion > YearsAtCompany': (df_clean['YearsSinceLastPromotion'] > df_clean['YearsAtCompany']).sum(),
    'YearsWithCurrManager > YearsAtCompany': (df_clean['YearsWithCurrManager'] > df_clean['YearsAtCompany']).sum(),
    'YearsAtCompany > TotalWorkingYears': (df_clean['YearsAtCompany'] > df_clean['TotalWorkingYears']).sum(),
}

print("Post-cleaning logical violations (all should be 0):")
for check, count in logic_checks.items():
    status = "✅" if count == 0 else "⚠️"
    print(f"  {status} {check}: {count}")
```

## 5. Zero-Variance Column Removal

`EmployeeCount`, `Over18`, and `StandardHours` were flagged in Phase 2 as constant across all records (confirmed there via `nunique() == 1`). Constant columns carry no analytical value and are removed here.

```python
zero_variance_cols = [c for c in df_clean.columns if df_clean[c].nunique() == 1]
print(f"Zero-variance columns confirmed: {zero_variance_cols}")

df_clean = df_clean.drop(columns=zero_variance_cols)
print(f"Shape after dropping zero-variance columns: {df_clean.shape}")
```

## 6. Outlier Assessment

Using the IQR method (1.5×IQR beyond Q1/Q3) on the key continuous variables. These are **flagged for awareness, not removed** — in HR data, extreme values (long tenure, high income, many prior employers) are typically legitimate rather than data errors, and removing them would bias attrition analysis against exactly the employee segments most worth understanding.

```python
outlier_cols = ['MonthlyIncome', 'DistanceFromHome', 'TotalWorkingYears', 'YearsAtCompany', 'NumCompaniesWorked']

outlier_summary = []
for col in outlier_cols:
    q1, q3 = df_clean[col].quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_outliers = ((df_clean[col] < lower) | (df_clean[col] > upper)).sum()
    outlier_summary.append({
        'Column': col, 'Lower Bound': round(lower, 1), 'Upper Bound': round(upper, 1),
        'Outlier Count': n_outliers, 'Outlier %': round(n_outliers / len(df_clean) * 100, 2)
    })

pd.DataFrame(outlier_summary)
```

# Phase 4 — Business Questions & Exploratory Data Analysis

This phase answers the seven business objectives defined in Phase 1, in the same order, using the cleaned dataset from Phase 3.

**A framing principle used throughout this phase:** every comparison below is shown as an **attrition rate (%)**, not a raw headcount of leavers. A group with more people will always produce a bigger raw leaver count even if its underlying risk is average or below — rate is what makes groups comparable to each other and to the company baseline. Every chart includes a reference line at the overall baseline rate so it's immediately clear which segments sit above or below it.

All charts are built with Plotly for interactivity (hover for exact values and headcounts).

```python
import plotly.express as px
import plotly.graph_objects as go

BASELINE = round((df_clean['Attrition'] == 'Yes').mean() * 100, 2)
print(f"Company-wide baseline attrition rate: {BASELINE}%")


def attrition_rate_table(dataframe, col):
    """Return attrition rate (%) and headcount per category of `col`."""
    rate = dataframe.groupby(col)['Attrition'].apply(lambda s: round((s == 'Yes').mean() * 100, 2))
    count = dataframe.groupby(col).size()
    return pd.DataFrame({'Attrition Rate (%)': rate, 'Headcount': count}).reset_index()


def plot_attrition_rate(dataframe, col, title, category_order=None, height=420, horizontal=False):
    """Bar chart of attrition rate by category, benchmarked against the company baseline."""
    t = attrition_rate_table(dataframe, col)
    if category_order:
        t[col] = pd.Categorical(t[col], categories=category_order, ordered=True)
        t = t.sort_values(col)
    else:
        t = t.sort_values('Attrition Rate (%)', ascending=False)

    if horizontal:
        fig = px.bar(t, y=col, x='Attrition Rate (%)', text='Attrition Rate (%)', orientation='h',
                      hover_data=['Headcount'], title=title, height=height,
                      color='Attrition Rate (%)', color_continuous_scale='Reds')
        fig.add_vline(x=BASELINE, line_dash='dash', line_color='gray')
        fig.update_traces(texttemplate='%{text}%', textposition='outside')
        fig.update_yaxes(categoryorder='total ascending')
    else:
        fig = px.bar(t, x=col, y='Attrition Rate (%)', text='Attrition Rate (%)',
                      hover_data=['Headcount'], title=title, height=height,
                      color='Attrition Rate (%)', color_continuous_scale='Reds')
        fig.add_hline(y=BASELINE, line_dash='dash', line_color='gray',
                       annotation_text=f'Baseline {BASELINE}%', annotation_position='top left')
        fig.update_traces(texttemplate='%{text}%', textposition='outside')

    fig.update_layout(showlegend=False, coloraxis_showscale=False)
    return fig
```

## Objective 1 — Measure Workforce and Attrition Performance

**Business question:** What is the total workforce, how many employees left, and what is the overall attrition rate?

## Objective 7 — Identify Workforce Segments for Management Attention

**Business question:** Combining the findings above, which employee groups show the highest attrition and may need targeted attention?

Since **OverTime** produced the single largest gap of any factor examined (Objective 4), it's worth checking whether it compounds with the organizational risk found in Objective 3 (Job Role).

```python
pivot = df_clean.pivot_table(index='JobRole', columns='OverTime', values='Attrition',
                              aggfunc=lambda s: round((s == 'Yes').mean() * 100, 1))
pivot = pivot.sort_values('Yes', ascending=False)
fig = px.imshow(pivot, text_auto=True, color_continuous_scale='Reds', height=480, aspect='auto',
                 title='Attrition Rate (%): Job Role × OverTime',
                 labels=dict(color='Attrition %'))
fig.show()
```

**Finding — the single highest-risk segment in the dataset:** **Sales Representatives who work overtime leave at 66.7%** — two out of every three. That's more than 4x the company baseline, and nearly double the Sales Representative role's own already-elevated average (39.8%). The sample size for this exact cell is modest (24 employees), so treat the precise number with some caution, but the direction and scale of the effect are consistent with everything else found in this phase (Sales Representative is the highest-risk role; OverTime is the strongest single factor) — this is a real compounding effect, not noise.

```python
dims = ['AgeGroup', 'Gender', 'MaritalStatus', 'EducationField', 'Department', 'JobRole', 'JobLevel',
        'OverTime', 'BusinessTravel', 'WorkLifeBalance', 'SalarySlab', 'JobSatisfaction',
        'EnvironmentSatisfaction', 'JobInvolvement', 'RelationshipSatisfaction', 'StockOptionLevel']

rows = []
for col in dims:
    rate = df_clean.groupby(col)['Attrition'].apply(lambda s: (s == 'Yes').mean() * 100)
    count = df_clean.groupby(col).size()
    for cat in rate.index:
        if count[cat] >= 50:  # exclude small-sample segments from the ranking
            rows.append({
                'Dimension': col, 'Segment': str(cat),
                'Attrition Rate (%)': round(rate[cat], 1), 'Headcount': count[cat],
                'vs. Baseline (pts)': round(rate[cat] - BASELINE, 1)
            })

top_segments = pd.DataFrame(rows).sort_values('Attrition Rate (%)', ascending=False).head(10).reset_index(drop=True)

fig = px.bar(top_segments, x='Attrition Rate (%)', y='Segment', orientation='h', color='Dimension',
             hover_data=['Headcount', 'vs. Baseline (pts)'], height=480,
             title='Top 10 Highest-Attrition Segments (min. 50 employees)')
fig.add_vline(x=BASELINE, line_dash='dash', line_color='gray', annotation_text=f'Baseline {BASELINE}%')
fig.update_yaxes(categoryorder='total ascending')
fig.show()

top_segments
```

**Finding:** ranking every examined segment (restricted to groups of 50+ employees, to avoid small-sample noise) confirms the same handful of factors repeat at the top: **Job Role (Sales Representative), Age (18–25), Job Involvement (lowest rating), Work-Life Balance (lowest rating), and OverTime** each independently push attrition to roughly double the baseline or beyond.

## 📌 Phase 4 Summary & Transition to Phase 5

### Key findings, ranked by strength of effect:
1. **OverTime** is the single strongest individual factor — a 3x gap (30.5% vs. 10.4%), and it compounds with role: Sales Representatives on overtime hit 66.7% attrition.

3. **Compensation** is a strong, clean gradient — the lowest salary band leaves at nearly 6x the rate of the highest.
4. **Sales Representative** is the highest-risk job role (39.8%) — more than double its own department's average.
5. **Job Involvement** is the cleanest satisfaction-style gradient (33.7% → 9.0%), stronger than Job Satisfaction itself.
6. One genuinely counterintuitive result: attrition is *highest*, not lowest, among employees promoted within the last year — worth further investigation rather than assuming "more promotions" is a simple retention fix.
