## 1.6 KPI Definitions

### A. Late Delivery Rate

Measures the percentage of eligible orders that are classified as late.

**Conceptual definition:**

Late Delivery Rate = Late Orders / Total Eligible(Arri) Orders × 100

**Business question:**

> How significant is the company's delivery-risk problem?

**Diagnostic dimensions:**

- Time
- Market
- Region
- Shipping Mode
- Product Category
- Seller
- Customer Segment

---

### B. Risk / Fraud Rate

Measures the proportion of transactions identified as fraudulent or risky according to the validated dataset definition.

**Conceptual definition:**

Risk/Fraud Rate = Risky or Fraudulent Orders / Total Eligible Orders × 100

**Business question:**

> How large is the company's transaction-risk exposure?

**Diagnostic dimensions:**

- Market
- Payment Type
- Customer
- Product
- Time
- Order characteristics

> The exact definition will depend on the validated meaning of the fraud/risk field.

---

### C. Total Profit

Measures the total profit generated during the analyzed period.

**Conceptual definition:**

Total Profit = Sum of Validated Order Profit

**Business question:**

> How much profit is the business generating?

---

### D. Profit Margin

Measures profitability relative to sales.

**Conceptual definition:**

Profit Margin = Total Profit / Total Sales × 100

**Business question:**

> Is revenue growth translating into profitable growth?

---

### E. Average Profit per Order

Measures the average profit generated by each eligible order.

**Conceptual definition:**

Average Profit per Order = Total Profit / Total Eligible Orders

**Business question:**

> How much profit does the average order generate?

---

### F. Discount Impact

The analysis will investigate whether increasing discount levels are associated with declining profitability.

**Business question:**

> At what point does discounting begin to significantly reduce profitability?

---

### G. Shipping Cost Impact

The analysis will investigate whether shipping costs are associated with lower order profitability.

**Business question:**

> Which shipping-related factors are contributing to profitability erosion?

### *2.7 Data Grain*

*What does one row represent?*
<br>

The identifier analysis indicates that the dataset contains **180,519 unique Order Item IDs across 180,519 records**, meaning that `Order Item Id` uniquely identifies each transactional row.

At the same time, the dataset contains only **65,752 distinct Order IDs**, demonstrating that a single order can contain multiple order-item records.

The dataset also contains:

- **20,652 distinct customers**
- **65,752 distinct orders**
- **180,519 order items**
- **118 distinct products**

Therefore, the initial analytical grain of the dataset is:

> **One row represents one Order Item / Order Line.**

This distinction is critical for subsequent analysis because row count cannot be used directly as the number of orders.

For example:

- `COUNT(Order Item Id)` → Order-item volume
- `COUNT(DISTINCT Order Id)` → Order volume
- `COUNT(DISTINCT Customer Id)` → Customer volume
- `COUNT(DISTINCT Product Card Id)` → Product volume

This grain definition will also be essential when designing the analytical data model and calculating KPIs in Power BI.

Furthermore, the maximum number of distinct `Product Card Id` values associated with a single `Order Item Id` is **1**, confirming that each Order Item is associated with exactly one product.

These findings support the following preliminary relationship structure:

**Customer → Order → Order Item → Product**

with one-to-many relationships from Customer to Order, Order to Order Item, and Product to Order Item.

The difference between the average Order Items per Order and the average distinct Products per Order indicates that some orders may contain multiple Order Item records associated with the same product. This relationship will be investigated further before finalizing the analytical model.

### *3.2 Completeness — Missing Values*

````python
missing_summary = (
    df.isna()
      .sum()
      .to_frame("Missing Values")
)

missing_summary["Missing %"] = (
    missing_summary["Missing Values"] / len(df) * 100
)

missing_summary = (
    missing_summary
    .sort_values("Missing Values", ascending=False)
)

missing_summary[missing_summary["Missing Values"] > 0]
````

### Missing Value Diagnosis & Handling Strategy

During our initial audit, four columns were identified with missing attributes:

1. **`Product Description` (100.00% Missing):** Entirely empty. Dropping to reduce structural clutter.
2. **`Order Zipcode` (86.24% Missing):** Extremely high sparsity. Dropping as geographic routing is already captured via `Order City`, `Order State`, and `Order Country`.
3. **`Customer Lname` (0.004% Missing):** Negligible missing count (8 records). Filling with `'Unknown'` to preserve row integrity.
4. **`Customer Zipcode` (0.002% Missing):** Imputing conditionally based on matching `Customer City` and `Customer Country` mode/first values.

#### Strategic Rationale:
Dropping high-sparsity columns prevents unnecessary memory load and model noise, while smart imputation for customer attributes preserves item-level transactions for accurate KPI calculation.

### *4.5 Missing Value Treatment*

````python
critical_columns = [
    "Order Id",
    "Order Item Id",
    "Customer Id",
    "Product Card Id",
    "order date (DateOrders)"
]

df_clean[critical_columns].isna().sum()
````

````python
# Create a copy of the raw data to maintain lineage
df_clean = df.copy()

print(f"Initial Shape: {df_clean.shape}")

# 1. Drop irrecoverable and high-sparsity columns
cols_to_drop = ["Product Description", "Order Zipcode"]
df_clean.drop(columns=cols_to_drop, inplace=True, errors="ignore")

# 2. Fill missing Customer Last Name with 'Unknown'
df_clean["Customer Lname"].fillna("Unknown", inplace=True)

# 3. Smart Imputation for Customer Zipcode based on Country and City
# First, fill using the mode zipcode for the same Customer City & Country
zip_lookup = (
    df_clean.groupby(["Customer Country", "Customer City"])["Customer Zipcode"]
    .transform(lambda x: x.mode()[0] if not x.mode().empty else np.nan)
)

df_clean["Customer Zipcode"] = df_clean["Customer Zipcode"].fillna(zip_lookup)

# If any Zipcode remains null, fill with 'Unknown' or overall Country fallback
df_clean["Customer Zipcode"].fillna("00000", inplace=True)

# 4. Verify that missing values are completely resolved
missing_after = df_clean[
    ["Customer Lname", "Customer Zipcode"]
].isnull().sum()

print("\n--- Missing Values Audit After Handling ---")
print(missing_after)
print(f"\nFinal Cleaned Shape: {df_clean.shape}")
````
